## Regression with Structured Data

Solve the California Housing Prices

a) Use 5-fold cross validation

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.datasets import fetch_california_housing

california_housing = fetch_california_housing(as_frame=True)

x = california_housing.data
y = california_housing.target

print(x)
print(y)

k = 5
kf = KFold(n_splits=k, shuffle=True, random_state=42)
```

b) Compare Linear Regression and Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression


model_decision_tree = DecisionTreeRegressor()
model_linear_regression = LinearRegression()
```

c) Evaluate the MSE and MAE (not using Power Transform)

Observations:
- MSE:

Mean Squared Error is useful for problems that we want to penalize big errors/outliners.

  - MAE:

Mean Absolute Error is useful for problems that we want a fair and less affected by extremes average.

```python
from sklearn.model_selection import cross_val_score
import numpy as np

scores_decision_tree_mse = cross_val_score(model_decision_tree, x, y, cv=kf, scoring='neg_mean_squared_error')
scores_decision_tree_mae = cross_val_score(model_decision_tree, x, y, cv=kf, scoring='neg_mean_absolute_error')
scores_linear_regression_mse = cross_val_score(model_linear_regression, x, y, cv=kf, scoring='neg_mean_squared_error')
scores_linear_regression_mae = cross_val_score(model_linear_regression, x, y, cv=kf, scoring='neg_mean_absolute_error')


avarage_decision_tree_mse = np.mean(scores_decision_tree_mse)
avarage_decision_tree_mae = np.mean(scores_decision_tree_mae)
avarage_linear_regression = np.mean(scores_linear_regression_mse)
avarage_linear_regression_mae = np.mean(scores_linear_regression_mae)

print("-Decision Tree:")
print("MSE: ", avarage_decision_tree_mse)
print("MAE: ", avarage_decision_tree_mae)

print("-Linear Regression:")
print("MSE: ", avarage_linear_regression)
print("MAE: ", avarage_linear_regression_mae)

```

d) Compare each model using power transform

```python
from sklearn.preprocessing import PowerTransformer

pt = PowerTransformer(method='yeo-johnson')
x = pt.fit_transform(x)

print(x)
print(y)

k = 5
kf = KFold(n_splits=k, shuffle=True, random_state=42)
```

Evaluate the MSE and MAE (using Power Transform)

```python
from sklearn.model_selection import cross_val_score
import numpy as np

scores_decision_tree_mse = cross_val_score(model_decision_tree, x, y, cv=kf, scoring='neg_mean_squared_error')
scores_decision_tree_mae = cross_val_score(model_decision_tree, x, y, cv=kf, scoring='neg_mean_absolute_error')
scores_linear_regression_mse = cross_val_score(model_linear_regression, x, y, cv=kf, scoring='neg_mean_squared_error')
scores_linear_regression_mae = cross_val_score(model_linear_regression, x, y, cv=kf, scoring='neg_mean_absolute_error')

avarage_decision_tree_mse_pt = np.mean(scores_decision_tree_mse)
avarage_decision_tree_mae_pt = np.mean(scores_decision_tree_mae)
avarage_linear_regression_pt = np.mean(scores_linear_regression_mse)
avarage_linear_regression_mae_pt = np.mean(scores_linear_regression_mae)

print("-Decision Tree:")
print("MSE: ", avarage_decision_tree_mse_pt)
print("MAE: ", avarage_decision_tree_mae_pt)

print("-Linear Regression:")
print("MSE: ", avarage_linear_regression_pt)
print("MAE: ", avarage_linear_regression_mae_pt)
```

Final Analysis:

```python
analysis = {
    "Model": ["Decision Tree", "Linear Regression"],
    "MSE": [avarage_decision_tree_mse, avarage_linear_regression],
    "MAE": [avarage_decision_tree_mae, avarage_linear_regression_mae],
    "MSE (Power Transform)": [avarage_decision_tree_mse_pt, avarage_linear_regression_pt],
    "MAE (Power Transform)": [avarage_decision_tree_mae_pt, avarage_linear_regression_mae_pt],
    "MSE (Z-scale)": [avarage_decision_tree_mse_z, avarage_linear_regression_z],
    "MAE (Z-scale)": [avarage_decision_tree_mae_z, avarage_linear_regression_mae_z],
    "MSE (Min-max Scale)": [avarage_decision_tree_mse_mm, avarage_linear_regression_mm],
    "MAE (Min-max Scale)": [avarage_decision_tree_mae_mm, avarage_linear_regression_mae_mm]
}

analysis_df = pd.DataFrame(analysis)
analysis_df.to_csv("Analysis.csv", index=False)
print(analysis_df)
```

## Analysis:

- Decision Tree MSE and MAE had small variations using transformation with Min-Max Scaler.
  - Best MSE: -0.511 (Original) ou -0.515 (Min-Max Scale)
  - Best MAE: -0.459 (Min-Max Scale)
- Linear Regression MSE and MAE showed a significant improvement with Power Transformation.
  - Best MSE: -0.507 (Z-Scale e Min-Max Scale)
  - Best MAE: -0.531 (Original) ou -0.535 (Z-Scale ou Min-Max-Scale)

The Decision Tree model did not show any significant impact from transformations, indicating that tree-based models are less sensitive to scaling. On the other hand, Linear Regression demonstrated clear improvements with feature transformations, confirming that linear models are more sensitive to scale and distribution.
