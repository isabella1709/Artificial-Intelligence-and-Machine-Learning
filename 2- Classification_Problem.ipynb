a) Use the Breast Cancer classification dataset


```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

data = load_breast_cancer(return_X_y=True, as_frame=True)
X = data[0]
y = data[1]
```

b) Use 5-fold cross validation

```python
from sklearn.model_selection import KFold
k = 5
kf = KFold(n_splits=k, shuffle=True, random_state=42)
```

c) Train and evaluate 3 different learning algorithms

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

model_decision_tree = DecisionTreeClassifier()

model_random_forest = RandomForestClassifier()

model_logistic_regression = LogisticRegression()

```

d) Compare matrics for the 3 methods

```python
from sklearn.model_selection import cross_val_score

scores_decision_tree = cross_val_score(model_decision_tree, X, y, cv=kf, scoring='f1')
scores_random_forest = cross_val_score(model_random_forest, X, y, cv=kf, scoring='f1')
scores_logistic_regression = cross_val_score(model_logistic_regression, X, y, cv=kf, scoring='f1')
```

e) Report experiment

```python
import numpy as np

average_f1_decision_tree = np.mean(scores_decision_tree)
average_f1_random_forest = np.mean(scores_random_forest)
average_f1_logistic_regression = np.mean(scores_logistic_regression)

# Exibição dos resultados formatados
print("- Decision Tree: \n")
print(f"F1 Score for each fold: {[round(score, 4) for score in scores_decision_tree]}")
print(f"Average F1 across {kf.get_n_splits()} folds: {average_f1_decision_tree:.2f}\n")

print("- Random Forest: \n")
print(f"F1 Score for each fold: {[round(score, 4) for score in scores_random_forest]}")
print(f"Average F1 across {kf.get_n_splits()} folds: {average_f1_random_forest:.2f}\n")

print("- Logistic Regression: \n")
print(f"F1 Score for each fold: {[round(score, 4) for score in scores_logistic_regression]}")
print(f"Average F1 across {kf.get_n_splits()} folds: {average_f1_logistic_regression:.2f}\n")
```
